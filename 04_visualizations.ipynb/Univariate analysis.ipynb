```python
import plotly.express as plt
import matplotlib.pyplot as plt
import plotly.express as plt
import plotly.express as px
import pandas as pd
import seaborn as sns
import numpy as np  
```

```python
# Load the movie data
df = pd.read_csv("movies_cleaned.csv")
```

```python
import matplotlib.pyplot as plt
```

```python
# Select top 10 movies by popularity
top10 = df.sort_values(by='popularity',ascending=False).head(10)

# Create pie chart
plt.figure(figsize=(10, 10))

plt.pie(top10['popularity'],labels=top10['title'],autopct='%1.1f%%',startangle=140)

plt.title('Top 10 Most Popular Movies', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# Select top 10 movies by vote average
top10 = df.sort_values(by='vote_average',ascending=False).head(10)

# Create donut chart
plt.figure(figsize=(10, 10))

plt.pie(top10['vote_average'],labels=top10['title'],autopct='%1.1f%%',startangle=140,pctdistance=0.75,
    wedgeprops=dict(width=0.4,edgecolor='white',linewidth=1.5 ))

plt.title(
    'Top 10 Movies by Vote Average',fontsize=14,fontweight='bold')
plt.tight_layout()
plt.show()
```

```python

keywords = pd.read_csv("movie_keywords_cleaned.csv")
movies = pd.read_csv("movies_cleaned.csv")

df = keywords.merge(movies, on="movie_id")

result = df.groupby("keyword_name")["budget"].sum()
result = result.sort_values(ascending=False).head(10)

plt.figure(figsize=(12, 6))
plt.bar(result.index, result.values)

plt.title("💰 Top 10 Keywords by Budget")
plt.xlabel("Keyword")
plt.ylabel("Total Budget")

plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

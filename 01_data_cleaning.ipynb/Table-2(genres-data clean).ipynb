```python
#check duplicate rows
print("Duplicate rows:", df_genres.duplicated().sum())
```

```python
# 1. Text normalization & whitespace removal
df_genres['genre_name'] = df_genres['genre_name'].astype(str).str.strip() 
```

```python
df_genres["genre_name"] = ( df_genres["genre_name"] .fillna("Unknown") .astype(str) .str.strip() .str.title() )
```

```python
df_genres
```

```python
#2. Remove duplicate genre IDs
df_genres_clean = df_genres.drop_duplicates(subset=['genre_id'], keep='first')
```

```python
df_genres_clean
```

```python
# Display cleaned data
print(df_genres.to_string(index=False))
```

```python
# 3. save cleaned dataset
df_genres_clean.to_csv('genres_cleaned.csv', index=False)
print("Genres dataset cleaned successfully!")
```

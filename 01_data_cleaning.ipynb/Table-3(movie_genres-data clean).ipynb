```python
# 2. Convert datatypes
df_mg["movie_id"] = df_mg["movie_id"].astype(int)
df_mg["genre_id"] = df_mg["genre_id"].astype(int)
```

```python
#check duplicate rows
print("Duplicate rows:", df_mg.duplicated().sum())
```

```python
# Remove duplicate mappings
df_mg_clean = (df_mg.drop_duplicates().reset_index(drop=True))
```

```python
df_mg_clean
```

```python
# 3. Load primary tables
df_movies = pd.read_csv('movies_cleaned.csv')
df_genres = pd.read_csv('genres_cleaned.csv')
```

```python
# 2. Validate foreign keys against referenced tables
valid_movies = set(df_movies['movie_id'])
valid_genres = set(df_genres['genre_id'])
```

```python
# 4. Filter out any invalid IDs
df_mg_clean = df_mg_clean[
    df_mg_clean['movie_id'].isin(valid_movies) & 
    df_mg_clean['genre_id'].isin(valid_genres)
]
```

```python
# 5.  cleaned dataset
df_mg_clean.to_csv('movie_genres_cleaned.csv', index=False)
print("Movie_genres dataset cleaned successfully!")
```

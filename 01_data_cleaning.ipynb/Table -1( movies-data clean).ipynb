```python
# Load dataset
df = pd.read_csv('movies.csv')
```

```python
# Copy dataframe
df_clean = df.copy()
print(df_clean.head())
```

```python
# 2. Check duplicate records
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Remove duplicates
df_clean = df.drop_duplicates()
```

```python
df_clean
```

```python
print("Duplicates after:", df_clean.duplicated().sum())
```

```python
# 2. Clean string columns
text_cols = df_clean.select_dtypes(include=["str"]).columns

for col in text_cols:
    df_clean[col] = df_clean[col].str.strip()
```

```python
df_clean[col]
```

```python
# 3. Check missing values
print("\nMissing values:")
print(df_clean.isnull().sum())
```

```python
# 5. Convert release_date to datetime
df_clean["release_date"] = pd.to_datetime(
    df_clean["release_date"],errors="coerce")
```

```python
df_clean
```

```python
# 6. Check invalid numeric values
print("\nNegative values:")
print("Budget:", (df_clean["budget"] < 0).sum())
print("Revenue:", (df_clean["revenue"] < 0).sum())
print("Runtime:", (df_clean["runtime"] < 0).sum())
```

```python
# 7. Replace invalid negative values with NaN
df_clean.loc[df_clean["budget"] < 0, "budget"] = np.nan
df_clean.loc[df_clean["revenue"] < 0, "revenue"] = np.nan
df_clean.loc[df_clean["runtime"] < 0, "runtime"] = np.nan
```

```python
# 8. Fill numeric missing values with median
numeric_cols = ["budget", "revenue", "runtime"]

for col in numeric_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())
```

```python
df_clean[col]
```

```python
# 9. Feature Engineering: Profit
df_clean["profit"] = df_clean["revenue"] - df_clean["budget"]
print(df_clean['profit'])
```

```python
# 10. Feature Engineering: ROI (%)
df_clean["roi_pct"] = np.where(df_clean["budget"] > 0,
    ((df_clean["revenue"] - df_clean["budget"]) / df_clean["budget"]) * 100,np.nan)
print(df_clean['roi_pct'])
```

```python
# Round ROI
df_clean["roi_pct"] = df_clean["roi_pct"].round(2) 
print(df_clean['roi_pct'])   
```

```python
# 11. Reset index
df_clean = df_clean.reset_index(drop=True)
print(df_clean)
```

```python

# 12. Final information
print("\nFinal shape:", df_clean.shape)

print("\nFinal data types:")
print(df_clean.dtypes)

print("\nFinal missing values:")
print(df_clean.isnull().sum())
print(df_clean.head())
```

```python
 #Export cleaned data
df_clean.to_csv("movies_cleaned.csv", index=False)

print("\nCleaned CSV exported successfully!")
```

```python
import pandas as pd
```

```python
# 1. Load crew and reference dataset
df_crew = pd.read_csv('crew.csv')
df_movies = pd.read_csv('movies_cleaned.csv')
```

```python
# 2. clean text columns
text_cols = ['person_name', 'job', 'department']
for col in text_cols:
    df_crew[col] = df_crew[col].astype(str).str.strip()
```

```python
#check duplicate rows
print("Duplicate rows:", df_crew.duplicated().sum())
```

```python
# 3. Drop duplicate records
df_crew_clean = df_crew.drop_duplicates(keep='first').copy()
```

```python
df_crew_clean
```

```python
df_crew_clean.shape
```

```python
# 4. Filter for valid movie_ids
valid_movies = set(df_movies['movie_id'])
df_crew_clean = df_crew_clean[df_crew_clean['movie_id'].isin(valid_movies)]
```

```python
# 5. Export cleaned CSV
df_crew_clean.to_csv('crew_cleaned.csv', index=False)
```
